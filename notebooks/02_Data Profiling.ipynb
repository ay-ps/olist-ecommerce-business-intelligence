```python
customers = pd.read_csv("../Data/raw/olist_customers_dataset.csv")
orders_items = pd.read_csv("../Data/raw/olist_order_items_dataset.csv", sep=";")
payments = pd.read_csv("../Data/raw/olist_order_payments_dataset.csv")
reviews = pd.read_csv("../Data/raw/olist_order_reviews_dataset.csv")
products = pd.read_csv("../Data/raw/olist_products_dataset.csv")
sellers = pd.read_csv("../Data/raw/olist_sellers_dataset.csv")
geolocations = pd.read_csv("../Data/raw/olist_geolocation_dataset.csv")
translations = pd.read_csv("../Data/raw/product_category_name_translation.csv", sep=";")
```

```python
summary = []

datasets = {
    'Orders': orders,
    'Customers': customers,
    'Order Items': orders_items,
    'Payments': payments,
    'Reviews': reviews,
    'Products': products,
    'Sellers': sellers,
    'Geolocations': geolocations,
    'Translations': translations
}

for name, df in datasets.items():
    summary.append({
        'Dataset': name,
        'Rows': df.shape[0],
        'Columns': df.shape[1],
        'Missing Values': df.isnull().sum().sum(),
        'Duplicate Rows': df.duplicated().sum()
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
profiling = []
for name, df in datasets.items():
    profiling.append({
        'Dataset': name,
        'Rows': df.shape[0],
        'Columns': df.shape[1],
        'missing_values': df.isnull().sum().to_dict(),
        'duplicate_rows': df.duplicated().sum(),

    })

profiling_df = pd.DataFrame(profiling)
profiling_df
```
